# Postgrad Study Engine — One-Click 14-Day Demo (`fictional-14-days`)

**Long-term memory engine for AI tutors (Kaoyan / Postgraduate Exam & Self-Study).**

Run the cell below to execute the built-in 14-day fictional study timeline (`examples/courses/fictional-14-days.json` + `limits-and-recall.json`) in an isolated temporary directory:
- **Zero dependencies / Zero API Key required** (pure Python 3 standard library)
- **Zero dirty files** (runs inside a temporary workspace and cleans up after rendering)
- **Outputs**: Full CLI study brief, memory model summary, and interactive HTML report (`demo.html`)

```python
import os
import shutil
import subprocess
import sys
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory(prefix="postgrad-colab-") as tmp_root:
    tmp_path = Path(tmp_root)
    # Use current checkout if already inside the repo; otherwise shallow-clone into temp
    if Path("scripts/study.py").is_file() and Path("examples/courses/fictional-14-days.json").is_file():
        engine_dir = Path.cwd()
    else:
        engine_dir = tmp_path / "postgrad-study-engine"
        subprocess.run(
            ["git", "clone", "--depth", "1", "https://github.com/wysking/postgrad-study-engine.git", str(engine_dir)],
            check=True,
        )

    demo_vault = tmp_path / "fictional-14-days-vault"
    completed = subprocess.run(
        [sys.executable, "-m", "scripts.study", "demo", "--vault", str(demo_vault)],
        cwd=str(engine_dir),
        capture_output=True,
        text=True,
        encoding="utf-8",
        check=True,
    )
    print("=== CLI Study Brief (fictional-14-days) ===\n")
    print(completed.stdout)

    report_html = (demo_vault / "reports" / "demo.html").read_text(encoding="utf-8")
    try:
        from IPython.display import HTML, display
        display(HTML(report_html))
    except ImportError:
        print("=== Generated HTML Report Length:", len(report_html), "chars ===")
```
